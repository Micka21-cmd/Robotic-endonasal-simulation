```python
from pathlib import Path
import re
import numpy as np
import pandas as pd

import matplotlib.pyplot as plt

# Use the variables already defined in the notebook
files = [Path(p) for p in csv_files]

def read_csv_auto(path):
    return pd.read_csv(path, sep=None, engine="python")

def clean_columns(df):
    df = df.copy()
    df.columns = [
        re.sub(r"[^a-zA-Z0-9]+", "_", str(c).strip()).strip("_").lower()
        for c in df.columns
    ]
    return df

def find_time_column(df):
    candidates = [c for c in df.columns if any(x in c for x in ["time", "timestamp", "step"])]
    return candidates[0] if candidates else None

def analyze_file(path):
    df = clean_columns(read_csv_auto(path))
    numeric = df.select_dtypes(include=np.number)

    print(f"\n{'=' * 80}\n{path.name}\n{'=' * 80}")
    print(f"Shape: {df.shape}")
    print("\nColumns and types:")
    print(df.dtypes)
    print("\nMissing values:")
    print(df.isna().sum())
    print("\nDescriptive statistics:")
    display(df.describe(include="all").T)

    time_col = find_time_column(df)
    if time_col and pd.api.types.is_numeric_dtype(df[time_col]):
        duration = df[time_col].max() - df[time_col].min()
        print(f"\nDetected time column: {time_col}")
        print(f"Simulation duration: {duration:.4g}")

    if not numeric.empty:
        print("\nNumeric variability:")
        variability = pd.DataFrame({
            "mean": numeric.mean(),
            "std": numeric.std(),
            "min": numeric.min(),
            "max": numeric.max(),
            "range": numeric.max() - numeric.min()
        }).sort_values("std", ascending=False)
        display(variability)

        ax = numeric.plot(
            subplots=True,
            layout=(-1, 2),
            figsize=(14, max(4, 3 * ((len(numeric.columns) + 1) // 2))),
            title=path.stem
        )
        plt.tight_layout()
        plt.show()

    return df

# Analyze the two files independently
data = {path.name: analyze_file(path) for path in files}

# Build a file-level comparison matrix
rows = []

for filename, df in data.items():
    numeric = df.select_dtypes(include=np.number)
    time_col = find_time_column(df)

    row = {
        "file": filename,
        "rows": len(df),
        "numeric_columns": len(numeric.columns),
        "missing_values": int(df.isna().sum().sum()),
    }

    if time_col and pd.api.types.is_numeric_dtype(df[time_col]):
        row["duration"] = df[time_col].max() - df[time_col].min()
    else:
        row["duration"] = np.nan

    # Useful aggregate indicators for robotics logs
    for col in numeric.columns:
        name = col.lower()
        if any(k in name for k in ["error", "deviation", "distance", "energy", "torque"]):
            row[f"{col}_mean_abs"] = numeric[col].abs().mean()
        elif any(k in name for k in ["success", "reward", "accuracy", "score"]):
            row[f"{col}_mean"] = numeric[col].mean()

    rows.append(row)

decision_matrix = pd.DataFrame(rows).set_index("file")
print("\nDecision matrix:")
display(decision_matrix.T)

# Normalize comparable metrics to [0, 1]
# Lower values are preferred for error, missing data, duration, energy, etc.
normalized = decision_matrix.select_dtypes(include=np.number).copy()

for column in normalized.columns:
    values = normalized[column]
    if values.nunique(dropna=True) <= 1:
        normalized[column] = 1.0
        continue

    lower_is_better = any(
        key in column.lower()
        for key in ["error", "deviation", "distance", "energy", "torque",
                    "missing", "duration"]
    )

    scaled = (values - values.min()) / (values.max() - values.min())
    normalized[column] = 1 - scaled if lower_is_better else scaled

print("\nNormalized decision matrix (1 = best):")
display(normalized.T)

if not normalized.empty:
    normalized["overall_score"] = normalized.mean(axis=1)
    print("\nOverall ranking:")
    display(normalized[["overall_score"]].sort_values("overall_score", ascending=False))
```
